### CELL 2.3 — Bangun `ALL_TRIALS` dari `meta_files`

````python
# ───────────────────────────────────────────────────────────────
#  Gunakan variabel `meta_files` yang SUDAH ADA dari TAHAP 1.
#  Bangun list of dict ALL_TRIALS tanpa menyentuh dataset.
# ───────────────────────────────────────────────────────────────

def build_trial_list(meta_files_list: list) -> list:
    """
    Dari list path meta.json, bangun ALL_TRIALS.
    Setiap elemen berisi:
      - meta_path   : path ke *_meta.json
      - data_path   : path ke *_processed_data.txt
      - group       : 'CVA' atau 'HS'
      - subject_id  : e.g. 'CVA_1'
      - trial_id    : e.g. 'CVA_1_1'
    """
    trials = []
    for meta_path in meta_files_list:
        meta_p = Path(meta_path)
        # Derive data path: ganti _meta.json -> _processed_data.txt
        data_path = str(meta_p).replace('_meta.json', '_processed_data.txt')
        if not os.path.exists(data_path):
            # Coba cari di folder yang sama
            candidates = list(meta_p.parent.glob('*_processed_data.txt'))
            if not candidates:
                continue
            data_path = str(candidates[0])

        # Tentukan group dari path
        parts = meta_p.parts
        group = 'CVA' if any('CVA' in p and not '_' in p.replace('CVA','') for p in parts) else 'HS'
        # Lebih robust: cek apakah 'CVA' atau 'HS' ada sebagai folder di path
        group = 'unknown'
        for p in parts:
            if p == 'CVA':
                group = 'CVA'; break
            if p == 'HS':
                group = 'HS'; break

        trial_id   = meta_p.parent.name    # e.g. CVA_1_1
        subject_id = meta_p.parent.parent.name  # e.g. CVA_1

        trials.append({
            'meta_path' : str(meta_path),
            'data_path' : str(data_path),
            'group'     : group,
            'subject_id': subject_id,
            'trial_id'  : trial_id,
        })
    return trials

ALL_TRIALS = build_trial_list(meta_files)

# Verifikasi
from collections import Counter
gc = Counter(t['group'] for t in ALL_TRIALS)
print(f"ALL_TRIALS dibangun dari meta_files:")
print(f"  Total : {len(ALL_TRIALS)}")
print(f"  CVA   : {gc.get('CVA', 0)}")
print(f"  HS    : {gc.get('HS', 0)}")
print(f"\nContoh 3 trials:")
for t in ALL_TRIALS[:3]:
    print(f"  {t['trial_id']:15s} | group={t['group']} | data_exists={os.path.exists(t['data_path'])}")
````

### CELL 2.4 — Fungsi Utilitas (Loader, Filter, Column Selector)

````python
# ═══════════════════════════════════════════════════════════════
#  UTILITY FUNCTIONS
#  Semua modular, tidak hardcode nama kolom
# ═══════════════════════════════════════════════════════════════

# ── 1. Flexible TXT loader ────────────────────────────────────
def load_txt_flexible(filepath: str) -> pd.DataFrame:
    """
    Baca processed_data.txt dengan auto-detect delimiter.
    Coba tab, comma, semicolon, whitespace.
    """
    for sep in ['	', ',', ';', r'\s+']:
        try:
            df = pd.read_csv(filepath, sep=sep, engine='python',
                             on_bad_lines='skip')
            if df.shape[1] >= 3:
                return df
        except Exception:
            continue
    raise ValueError(f"Gagal parsing: {filepath}")

# ── 2. Clean column names ─────────────────────────────────────
def clean_columns(df: pd.DataFrame) -> pd.DataFrame:
    df.columns = [str(c).strip().replace(' ', '_') for c in df.columns]
    return df

# ── 3. Select IMU channels ────────────────────────────────────
SENSOR_KEYWORDS = ['acc', 'gyr', 'gyro', 'accel', 'mag']

def select_imu_channels(df: pd.DataFrame) -> tuple:
    """
    Pilih kolom numerik yang relevan dengan IMU.
    Prioritas: kolom mengandung kata kunci sensor (acc/gyr).
    Fallback: semua kolom numerik.
    Return: (array (T,C), list kolom names)
    """
    numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    sensor_cols  = [c for c in numeric_cols
                    if any(kw in c.lower() for kw in SENSOR_KEYWORDS)]
    chosen = sensor_cols if len(sensor_cols) >= 3 else numeric_cols
    if len(chosen) < 1:
        raise ValueError("Tidak cukup kolom numerik.")
    return df[chosen].fillna(0).values.astype(np.float32), chosen

# ── 4. Replace nan/inf ────────────────────────────────────────
def safe_arr(arr: np.ndarray) -> np.ndarray:
    return np.where(np.isfinite(arr), arr, 0.0).astype(np.float32)

# ── 5. Get Lower Back axis columns ────────────────────────────
def get_lb_axis_map(df: pd.DataFrame) -> dict:
    """
    Cari kolom Lower Back untuk ACC dan GYR.
    Return dict {'CC':col, 'ML':col, 'AP':col, 'GyrX':col, ...}
    Fallback ke 3 kolom acc pertama jika tidak ada label LB eksplisit.
    """
    cols   = df.columns.tolist()
    num_c  = df.select_dtypes(np.number).columns.tolist()
    result = {}

    # Coba match dari config (AXIS_CC, AXIS_ML, AXIS_AP)
    if AXIS_CC in df.columns: result['CC'] = AXIS_CC
    if AXIS_ML in df.columns: result['ML'] = AXIS_ML
    if AXIS_AP in df.columns: result['AP'] = AXIS_AP

    # Cari kolom LB / lower_back secara eksplisit
    lb_acc = [c for c in num_c if 'lb' in c.lower() and 'acc' in c.lower()]
    lb_gyr = [c for c in num_c if 'lb' in c.lower() and ('gyr' in c.lower() or 'gyro' in c.lower())]
    all_acc= [c for c in num_c if 'acc' in c.lower()]
    all_gyr= [c for c in num_c if 'gyr' in c.lower() or 'gyro' in c.lower()]

    acc_pool = lb_acc if lb_acc else all_acc
    gyr_pool = lb_gyr if lb_gyr else all_gyr

    if 'CC' not in result and len(acc_pool) >= 1: result['CC'] = acc_pool[0]
    if 'ML' not in result and len(acc_pool) >= 2: result['ML'] = acc_pool[1]
    if 'AP' not in result and len(acc_pool) >= 3: result['AP'] = acc_pool[2]
    if len(gyr_pool) >= 1: result['GyrX'] = gyr_pool[0]
    if len(gyr_pool) >= 2: result['GyrY'] = gyr_pool[1]
    if len(gyr_pool) >= 3: result['GyrZ'] = gyr_pool[2]

    return result

# ── 6. Butterworth low-pass filter ───────────────────────────
def butter_lowpass(data: np.ndarray, cutoff=20.0, fs=100.0, order=4):
    nyq = fs / 2.0
    b, a = butter(order, min(cutoff/nyq, 0.99), btype='low', analog=False)
    if data.ndim == 1:
        return filtfilt(b, a, data)
    return np.apply_along_axis(lambda c: filtfilt(b, a, c), 0, data)

print("Utility functions siap.")
````

### CELL 2.5 — Load Satu Trial (Demo)

````python
def load_trial(trial_info: dict) -> dict:
    """
    Load satu trial: baca meta.json + processed_data.txt.
    Input : dict dari ALL_TRIALS
    Output: dict {meta, signal (T,C), columns, df, trial_id, group}
    Tidak mengubah dataset apapun.
    """
    # Load meta
    with open(trial_info['meta_path'], 'r') as f:
        meta = json.load(f)

    # Load signal
    df = load_txt_flexible(trial_info['data_path'])
    df = clean_columns(df)
    df = df.dropna(how='all').reset_index(drop=True)

    X, chosen_cols = select_imu_channels(df)
    X = safe_arr(X)

    return {
        'meta'    : meta,
        'signal'  : X,
        'columns' : chosen_cols,
        'df'      : df,
        'trial_id': trial_info['trial_id'],
        'group'   : trial_info['group'],
    }

# Demo load satu trial
if ALL_TRIALS:
    _demo = load_trial(ALL_TRIALS[0])
    print(f"Trial     : {_demo['trial_id']} | Group: {_demo['group']}")
    print(f"Signal    : shape={_demo['signal'].shape}  dtype={_demo['signal'].dtype}")
    print(f"Channels  : {_demo['columns']}")
    print(f"Freq      : {_demo['meta'].get('freq','?')} Hz")
    print(f"leftGaitEvents  (sample) : {_demo['meta'].get('leftGaitEvents',[])[:2]}")
    print(f"rightGaitEvents (sample) : {_demo['meta'].get('rightGaitEvents',[])[:2]}")
````

### CELL 2.6 — Label Generation dari meta.json

````python
def generate_labels(T: int, meta: dict,
                     tolerance: int = EVENT_TOLERANCE) -> np.ndarray:
    """
    Buat array label panjang T.
    Label: 0=None, 1=LHS, 2=LTO, 3=RHS, 4=RTO

    Format meta.json:
      leftGaitEvents[i]  = [LTO_idx, LHS_idx]
      rightGaitEvents[i] = [RTO_idx, RHS_idx]

    Setiap event diberi toleransi window ±tolerance sample.
    """
    labels = np.zeros(T, dtype=np.int64)

    def mark(label_val, idx):
        lo = max(0, idx - tolerance)
        hi = min(T, idx + tolerance + 1)
        labels[lo:hi] = label_val

    for pair in meta.get('leftGaitEvents', []):
        if len(pair) >= 2:
            lto_i, lhs_i = int(pair[0]), int(pair[1])
            if 0 <= lto_i < T: mark(2, lto_i)   # LTO = class 2
            if 0 <= lhs_i < T: mark(1, lhs_i)   # LHS = class 1

    for pair in meta.get('rightGaitEvents', []):
        if len(pair) >= 2:
            rto_i, rhs_i = int(pair[0]), int(pair[1])
            if 0 <= rto_i < T: mark(4, rto_i)   # RTO = class 4
            if 0 <= rhs_i < T: mark(3, rhs_i)   # RHS = class 3

    return labels

# Demo
if ALL_TRIALS:
    _T  = _demo['signal'].shape[0]
    _y  = generate_labels(_T, _demo['meta'])
    uni, cnt = np.unique(_y, return_counts=True)
    print(f"Label distribution — Trial {_demo['trial_id']} (T={_T}):")
    for u, c in zip(uni, cnt):
        print(f"  {CLASS_NAMES[u]:6s} (label={u}) : {c:6d} frames  ({100*c/_T:.2f}%)")
````

### CELL 2.7 — Preprocessing Signal

````python
def preprocess_signal(X: np.ndarray, freq: float = 100.0,
                       cutoff: float = 20.0) -> np.ndarray:
    """
    Preprocessing sinyal IMU:
      1. Replace NaN/Inf dengan 0
      2. Butterworth low-pass filter (cutoff=20 Hz)
    Normalisasi (z-score) dilakukan secara global setelah split.
    """
    X = safe_arr(X).copy()
    if X.shape[0] > 20:
        try:
            X = butter_lowpass(X, cutoff=cutoff, fs=freq)
        except Exception:
            pass   # lewati filter jika data terlalu pendek
    return X.astype(np.float32)

# StandardScaler — di-fit HANYA dari training set (diisi di CELL 2.9)
GLOBAL_SCALER = None

def fit_scaler(X_list: list) -> StandardScaler:
    """Fit StandardScaler dari list array (T,C)."""
    all_data = np.vstack(X_list)
    sc = StandardScaler()
    sc.fit(all_data)
    return sc

def apply_scaler(X: np.ndarray, scaler) -> np.ndarray:
    return scaler.transform(X).astype(np.float32)

print("Preprocessing functions siap.")
````

### CELL 2.8 — Sliding Window

````python
def make_windows(X: np.ndarray, y: np.ndarray,
                  window_size: int = WINDOW_SIZE,
                  stride: int = STRIDE) -> tuple:
    """
    Buat sliding window dari satu trial.
    X: (T, C)  →  X_wins: (N, window_size, C)
    y: (T,)    →  y_wins: (N, window_size)
    Trial lebih pendek dari window_size akan di-pad dengan nol.
    """
    T = X.shape[0]
    if T < window_size:
        pad = window_size - T
        X   = np.vstack([X, np.zeros((pad, X.shape[1]), dtype=np.float32)])
        y   = np.concatenate([y, np.zeros(pad, dtype=np.int64)])
        T   = window_size

    X_wins, y_wins = [], []
    for start in range(0, T - window_size + 1, stride):
        end = start + window_size
        X_wins.append(X[start:end])
        y_wins.append(y[start:end])

    return (np.array(X_wins, dtype=np.float32),
            np.array(y_wins, dtype=np.int64))

print(f"Sliding window siap: WINDOW_SIZE={WINDOW_SIZE}, STRIDE={STRIDE}")
````

### CELL 2.9 — Subject-Level Train/Val/Test Split & Load Semua Trial

````python
def subject_split(trials: list, train_r=0.70, val_r=0.15,
                   seed=SEED) -> tuple:
    """
    Split berdasarkan SUBJECT (bukan window) → cegah data leakage.
    CVA dan HS di-split terpisah agar distribusi terjaga.
    """
    np.random.seed(seed)
    tr, va, te = [], [], []

    for group in ['CVA', 'HS']:
        g_trials = [t for t in trials if t['group'] == group]
        subjects = sorted({t['subject_id'] for t in g_trials})
        np.random.shuffle(subjects)
        n      = len(subjects)
        n_tr   = int(n * train_r)
        n_va   = int(n * val_r)
        s_tr   = set(subjects[:n_tr])
        s_va   = set(subjects[n_tr:n_tr+n_va])
        s_te   = set(subjects[n_tr+n_va:])
        for t in g_trials:
            s = t['subject_id']
            if   s in s_tr: tr.append(t)
            elif s in s_va: va.append(t)
            else:           te.append(t)

    print(f"Split (subject-level): train={len(tr)}  val={len(va)}  test={len(te)}")
    return tr, va, te

TRAIN_TRIALS, VAL_TRIALS, TEST_TRIALS = subject_split(ALL_TRIALS)

# ── Load + preprocess semua trials ───────────────────────────
def load_all(trial_list: list, desc='Loading') -> list:
    data, errs = [], []
    for info in tqdm(trial_list, desc=desc):
        try:
            t    = load_trial(info)
            freq = float(t['meta'].get('freq', 100.0))
            X    = preprocess_signal(t['signal'], freq=freq)
            y    = generate_labels(X.shape[0], t['meta'])
            data.append({
                'X': X, 'y': y,
                'meta'    : t['meta'],
                'df'      : t['df'],
                'trial_id': info['trial_id'],
                'group'   : info['group'],
                'subject' : info['subject_id'],
            })
        except Exception as e:
            errs.append((info['trial_id'], str(e)))
    if errs:
        print(f"[WARN] {len(errs)} trial gagal load:")
        for tid, err in errs[:5]: print(f"  {tid}: {err}")
    return data

print("\nLoading training set...")
TRAIN_DATA = load_all(TRAIN_TRIALS, desc='Train')
print(f"  → {len(TRAIN_DATA)} trials")

# Fit global scaler HANYA dari training data
GLOBAL_SCALER = fit_scaler([d['X'] for d in TRAIN_DATA])
for d in TRAIN_DATA:
    d['X'] = apply_scaler(d['X'], GLOBAL_SCALER)

print("Loading val & test set...")
VAL_DATA  = load_all(VAL_TRIALS,  desc='Val')
TEST_DATA = load_all(TEST_TRIALS, desc='Test')
for d in VAL_DATA + TEST_DATA:
    d['X'] = apply_scaler(d['X'], GLOBAL_SCALER)

# Simpan scaler
with open(SCALER_PATH, 'wb') as f:
    pickle.dump(GLOBAL_SCALER, f)

N_CHANNELS = TRAIN_DATA[0]['X'].shape[1] if TRAIN_DATA else 6
print(f"\nInput channels  : {N_CHANNELS}")
print(f"Scaler saved    : {SCALER_PATH}")
````

````python
class GaitWindowDataset(Dataset):
    """
    Sliding window dataset dari list data trial.
    Setiap item: (X_window: FloatTensor (W,C), y_window: LongTensor (W,))
    """
    def __init__(self, data_list: list,
                 window_size=WINDOW_SIZE, stride=STRIDE):
        self.samples = []
        for d in data_list:
            X_w, y_w = make_windows(d['X'], d['y'], window_size, stride)
            for i in range(len(X_w)):
                self.samples.append((X_w[i], y_w[i]))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        X, y = self.samples[idx]
        return (torch.from_numpy(X).float(),
                torch.from_numpy(y).long())

train_ds = GaitWindowDataset(TRAIN_DATA)
val_ds   = GaitWindowDataset(VAL_DATA)
test_ds  = GaitWindowDataset(TEST_DATA)

print(f"Windows: train={len(train_ds)}  val={len(val_ds)}  test={len(test_ds)}")

# ── Class weights (inverse frequency) ────────────────────────
label_counts = np.zeros(NUM_CLASSES, dtype=np.float64)
for X_, y_ in train_ds:
    vals, cnts = torch.unique(y_, return_counts=True)
    for v, c in zip(vals.numpy(), cnts.numpy()):
        label_counts[int(v)] += int(c)

label_counts  = np.where(label_counts == 0, 1, label_counts)
class_weights = 1.0 / label_counts
class_weights = class_weights / class_weights.sum() * NUM_CLASSES
CLASS_WEIGHTS = torch.tensor(class_weights, dtype=torch.float32).to(device)

print("\nClass weights (inverse frequency):")
for i, w in enumerate(CLASS_WEIGHTS.cpu().numpy()):
    print(f"  {CLASS_NAMES[i]:6s}: count={int(label_counts[i]):9d}  weight={w:.4f}")

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
print(f"\nDataLoaders siap — batch_size={BATCH_SIZE}")
````

### CELL 2.11 — BiTCN–BiGRU–Cross Attention Model

Arsitektur berdasarkan Xu et al. (2025) *Bioengineering* 12(5):491:

```
Input (B, T, C)
→ BiTCN (forward + backward dilated TCN, concat)
→ BatchNorm + ReLU
→ BiGRU (2 layer)
→ Cross Attention (MultiheadAttention, Q=K=V)
→ LayerNorm + residual
→ Linear → (B, T, 5)
```


````python
# ── 1. Causal Dilated Convolution ────────────────────────────
class CausalConv1d(nn.Module):
    """Dilated causal 1D conv: tidak bocorkan info future."""
    def __init__(self, in_ch, out_ch, kernel_size, dilation):
        super().__init__()
        self.pad  = (kernel_size - 1) * dilation
        self.conv = nn.Conv1d(in_ch, out_ch, kernel_size,
                              padding=self.pad, dilation=dilation)

    def forward(self, x):               # x: (B, C, T)
        return self.conv(x)[:, :, :x.size(2)]   # trim padding

# ── 2. TCN Residual Block ─────────────────────────────────────
class TCNBlock(nn.Module):
    """
    Residual block:
      CausalConv → BN → ReLU → Dropout
      CausalConv → BN → ReLU → Dropout
      + residual (1x1 conv jika channel berubah)
    """
    def __init__(self, in_ch, out_ch, kernel_size, dilation,
                 dropout=0.2):
        super().__init__()
        self.conv1 = CausalConv1d(in_ch,  out_ch, kernel_size, dilation)
        self.conv2 = CausalConv1d(out_ch, out_ch, kernel_size, dilation)
        self.bn1   = nn.BatchNorm1d(out_ch)
        self.bn2   = nn.BatchNorm1d(out_ch)
        self.relu  = nn.ReLU()
        self.drop  = nn.Dropout(dropout)
        self.proj  = (nn.Conv1d(in_ch, out_ch, 1)
                      if in_ch != out_ch else nn.Identity())

    def forward(self, x):               # x: (B, C, T)
        res = self.proj(x)
        out = self.drop(self.relu(self.bn1(self.conv1(x))))
        out = self.drop(self.relu(self.bn2(self.conv2(out))))
        return self.relu(out + res)

# ── 3. Bidirectional TCN ──────────────────────────────────────
class BiTCN(nn.Module):
    """
    Forward TCN (dilation 1,2,4,...) + Backward TCN (flip), concat.
    Output dim = channels[-1] * 2
    """
    def __init__(self, in_ch, channels: list,
                 kernel_size=3, dropout=0.2):
        super().__init__()
        self.fwd = nn.ModuleList()
        self.bwd = nn.ModuleList()
        ch = in_ch
        for i, out_ch in enumerate(channels):
            dil = 2 ** i
            self.fwd.append(TCNBlock(ch, out_ch, kernel_size, dil, dropout))
            self.bwd.append(TCNBlock(ch, out_ch, kernel_size, dil, dropout))
            ch = out_ch
        self.out_channels = channels[-1] * 2

    def forward(self, x):               # x: (B, T, C_in)
        x_t  = x.transpose(1, 2)        # → (B, C_in, T)
        f    = x_t
        b    = torch.flip(x_t, dims=[2])
        for fb, bb in zip(self.fwd, self.bwd):
            f = fb(f)
            b = bb(b)
        b    = torch.flip(b, dims=[2])
        out  = torch.cat([f, b], dim=1) # (B, ch*2, T)
        return out.transpose(1, 2)      # (B, T, ch*2)

# ── 4. Full Model ─────────────────────────────────────────────
class BiTCN_BiGRU_CrossAttn(nn.Module):
    """
    BiTCN → BN+ReLU → BiGRU → CrossAttention → Linear
    Output: (B, T, num_classes)
    """
    def __init__(self, in_channels: int,
                 tcn_channels=TCN_CHANNELS,
                 tcn_kernel=TCN_KERNEL,
                 tcn_dropout=TCN_DROPOUT,
                 gru_hidden=GRU_HIDDEN,
                 gru_layers=GRU_LAYERS,
                 attn_heads=ATTN_HEADS,
                 attn_dropout=ATTN_DROPOUT,
                 num_classes=NUM_CLASSES):
        super().__init__()

        # BiTCN
        self.bitcn    = BiTCN(in_channels, tcn_channels,
                               tcn_kernel, tcn_dropout)
        tcn_dim       = self.bitcn.out_channels   # ch*2

        # BN + ReLU setelah BiTCN
        self.bn_tcn   = nn.BatchNorm1d(tcn_dim)
        self.relu     = nn.ReLU()

        # BiGRU
        self.bigru    = nn.GRU(
            input_size   = tcn_dim,
            hidden_size  = gru_hidden,
            num_layers   = gru_layers,
            batch_first  = True,
            bidirectional= True,
            dropout      = tcn_dropout if gru_layers > 1 else 0.0
        )
        gru_dim = gru_hidden * 2           # bidirectional output

        # Cross Attention (self-attention: Q=K=V = BiGRU output)
        self.cross_attn = nn.MultiheadAttention(
            embed_dim   = gru_dim,
            num_heads   = attn_heads,
            dropout     = attn_dropout,
            batch_first = True
        )
        self.attn_norm  = nn.LayerNorm(gru_dim)

        # Classifier head
        self.classifier = nn.Linear(gru_dim, num_classes)

    def forward(self, x):               # x: (B, T, C)
        # BiTCN
        tcn_out = self.bitcn(x)          # (B, T, tcn_dim)
        # BN: transpose → BN → transpose
        tcn_out = self.bn_tcn(
            tcn_out.transpose(1,2)
        ).transpose(1,2)
        tcn_out = self.relu(tcn_out)
        # BiGRU
        gru_out, _ = self.bigru(tcn_out) # (B, T, gru_dim)
        # Cross Attention (residual connection)
        attn_out, _ = self.cross_attn(gru_out, gru_out, gru_out)
        attn_out     = self.attn_norm(attn_out + gru_out)
        # Classifier
        return self.classifier(attn_out) # (B, T, num_classes)

# ── Instantiate ───────────────────────────────────────────────
model = BiTCN_BiGRU_CrossAttn(in_channels=N_CHANNELS).to(device)

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Model     : BiTCN_BiGRU_CrossAttn")
print(f"Params    : {n_params:,}")
print(f"In channels: {N_CHANNELS}")

# Verifikasi forward pass
_xb = torch.randn(2, WINDOW_SIZE, N_CHANNELS).to(device)
_out = model(_xb)
print(f"Forward pass OK: input={_xb.shape} → output={_out.shape}")
del _xb, _out
````

### CELL 2.12 — Training Loop

````python
def run_epoch(model, loader, optimizer, criterion,
              device, train=True):
    """Satu epoch training atau evaluasi."""
    model.train() if train else model.eval()
    total_loss = 0.0
    all_pred, all_true = [], []

    ctx = torch.enable_grad() if train else torch.no_grad()
    with ctx:
        for X_b, y_b in loader:
            X_b, y_b = X_b.to(device), y_b.to(device)
            logits   = model(X_b)                     # (B, T, C)
            loss     = criterion(
                logits.reshape(-1, NUM_CLASSES),
                y_b.reshape(-1))

            if train:
                optimizer.zero_grad()
                loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()

            total_loss += loss.item() * X_b.size(0)
            preds = logits.argmax(-1).reshape(-1).cpu().numpy()
            all_pred.append(preds)
            all_true.append(y_b.reshape(-1).cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)
    all_pred = np.concatenate(all_pred)
    all_true = np.concatenate(all_true)
    macro_f1 = f1_score(all_true, all_pred, average='macro',
                         zero_division=0,
                         labels=list(range(NUM_CLASSES)))
    return avg_loss, macro_f1, all_true, all_pred

# ── Setup ─────────────────────────────────────────────────────
criterion = nn.CrossEntropyLoss(weight=CLASS_WEIGHTS)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR,
                               weight_decay=WEIGHT_DECAY)


history       = defaultdict(list)
best_val_f1   = -1.0
patience_cnt  = 0
best_state    = None

print(f"Training — max {EPOCHS} epochs | early stop patience={PATIENCE}")
print("-" * 72)

for epoch in range(1, EPOCHS + 1):
    t0 = time.time()
    tr_loss, tr_f1, _, _   = run_epoch(model, train_loader,
                                        optimizer, criterion, device, train=True)
    va_loss, va_f1, _, _   = run_epoch(model, val_loader,
                                        optimizer, criterion, device, train=False)
    #scheduler.step(va_f1)
    dt = time.time() - t0

    history['train_loss'].append(tr_loss)
    history['val_loss'].append(va_loss)
    history['train_f1'].append(tr_f1)
    history['val_f1'].append(va_f1)

    tag = ''
    if va_f1 > best_val_f1:
        best_val_f1  = va_f1
        best_state   = deepcopy(model.state_dict())
        patience_cnt = 0
        torch.save(best_state, MODEL_PATH)
        tag = '  ✓ best'
    else:
        patience_cnt += 1

    print(f"Epoch {epoch:3d}/{EPOCHS} | "
          f"TrLoss={tr_loss:.4f} TrF1={tr_f1:.4f} | "
          f"VaLoss={va_loss:.4f} VaF1={va_f1:.4f} | "
          f"{dt:.1f}s{tag}")

    if patience_cnt >= PATIENCE:
        print(f"Early stopping at epoch {epoch}.")
        break

model.load_state_dict(torch.load(MODEL_PATH, map_location=device))
print(f"\nTraining selesai. Best val macro F1 = {best_val_f1:.4f}")
print(f"Model disimpan : {MODEL_PATH}")
````

### CELL 2.14 — Event Timestamp Extraction

````python
def extract_events_from_sequence(pred_labels: np.ndarray,
                                   freq: float = 100.0,
                                   min_distance: int = 20) -> dict:
    """
    Ubah label sequence → timestamp event per kelas.

    Langkah:
      1. Untuk tiap kelas (1–4), cari segmen berturut-turut (connected components)
      2. Ambil midpoint setiap segmen sebagai event index
      3. Hapus event yang jaraknya < min_distance (ambil yang pertama)

    Return:
      {'LHS': [idx,...], 'LTO': [idx,...], 'RHS': [idx,...], 'RTO': [idx,...]}
    """
    results = {'LHS': [], 'LTO': [], 'RHS': [], 'RTO': []}
    label_to_name = {1: 'LHS', 2: 'LTO', 3: 'RHS', 4: 'RTO'}

    for lv, name in label_to_name.items():
        mask = (pred_labels == lv).astype(np.int32)
        if mask.sum() == 0:
            continue

        labeled, n_comp = ndimage_label(mask)
        indices = []
        for cid in range(1, n_comp + 1):
            positions = np.where(labeled == cid)[0]
            indices.append(int(positions.mean()))

        # Filter: hapus event terlalu dekat
        if len(indices) > 1:
            indices.sort()
            kept = [indices[0]]
            for idx in indices[1:]:
                if idx - kept[-1] >= min_distance:
                    kept.append(idx)
            indices = kept

        results[name] = indices

    return results

def events_to_seconds(events: dict, freq: float) -> dict:
    return {k: [round(i / freq, 4) for i in v] for k, v in events.items()}

def parse_gt_events(meta: dict) -> dict:
    """Parse ground truth dari meta.json."""
    gt = {'LHS': [], 'LTO': [], 'RHS': [], 'RTO': []}
    for pair in meta.get('leftGaitEvents', []):
        if len(pair) >= 2:
            gt['LTO'].append(int(pair[0]))
            gt['LHS'].append(int(pair[1]))
    for pair in meta.get('rightGaitEvents', []):
        if len(pair) >= 2:
            gt['RTO'].append(int(pair[0]))
            gt['RHS'].append(int(pair[1]))
    return gt

print("extract_events_from_sequence() siap.")
````

### CELL 2.15 — Full Trial Inference (Overlap Averaging)

````python
def predict_full_trial(model, X: np.ndarray,
                        window_size: int = WINDOW_SIZE,
                        stride: int = STRIDE,
                        num_classes: int = NUM_CLASSES,
                        device=device) -> tuple:
    """
    Inference trial penuh dengan sliding window.
    Logits pada area overlap di-rata-rata (averaging).

    Input : X (T, C) — sudah di-scaled
    Output: (pred_labels (T,), logits_full (T, num_classes))
    """
    model.eval()
    T    = X.shape[0]
    T_pad = max(T, window_size)
    pad  = T_pad - T

    if pad > 0:
        X_pad = np.vstack([X, np.zeros((pad, X.shape[1]), np.float32)])
    else:
        X_pad = X

    logits_acc = np.zeros((T_pad, num_classes), np.float64)
    count_acc  = np.zeros(T_pad, np.float64)

    starts = list(range(0, T_pad - window_size + 1, stride))
    if not starts or starts[-1] + window_size < T_pad:
        starts.append(max(0, T_pad - window_size))

    with torch.no_grad():
        for s in starts:
            e   = s + window_size
            X_w = torch.from_numpy(X_pad[s:e]).unsqueeze(0).float().to(device)
            lg  = model(X_w).squeeze(0).cpu().numpy()  # (W, C)
            logits_acc[s:e] += lg
            count_acc[s:e]  += 1.0

    count_acc  = np.where(count_acc == 0, 1, count_acc)
    logits_full = (logits_acc / count_acc[:, None])[:T]
    pred_labels = logits_full.argmax(axis=1).astype(np.int64)

    return pred_labels, logits_full

# ── Demo pada satu trial test ─────────────────────────────────
if TEST_DATA:
    _td = TEST_DATA[0]
    _preds, _logits = predict_full_trial(model, _td['X'])
    _freq  = float(_td['meta'].get('freq', 100.0))
    _evts  = extract_events_from_sequence(_preds, freq=_freq)
    _gt    = parse_gt_events(_td['meta'])

    print(f"Demo trial : {_td['trial_id']}")
    print(f"Pred shape : {_preds.shape}")
    print(f"Events pred: { {k: len(v) for k,v in _evts.items()} }")
    print(f"Events GT  : { {k: len(v) for k,v in _gt.items()} }")
````

### CELL 2.16 — Event-Level MAE (Millisecond)

````python
def compute_event_mae(pred_ev: dict, gt_ev: dict,
                       freq: float = 100.0) -> dict:
    """
    Hitung MAE timing error (ms) per event type.
    Gunakan greedy nearest-neighbor matching antara pred dan GT.
    MAE (ms) = mean(|t_pred - t_true|) / freq * 1000
    """
    mae = {}
    for name in ['LHS', 'LTO', 'RHS', 'RTO']:
        pred = np.array(sorted(pred_ev.get(name, [])))
        gt   = np.array(sorted(gt_ev.get(name,   [])))

        if len(pred) == 0 or len(gt) == 0:
            mae[name] = np.nan
            continue

        errors, used = [], set()
        for p in pred:
            diffs = np.abs(gt - p).copy()
            for u in used: diffs[u] = 1e9
            j = int(np.argmin(diffs))
            if diffs[j] < 1e8:
                errors.append(abs(int(p) - int(gt[j])))
                used.add(j)

        mae[name] = np.mean(errors) / freq * 1000.0 if errors else np.nan

    valid       = [v for v in mae.values() if not np.isnan(v)]
    mae['overall'] = np.mean(valid) if valid else np.nan
    return mae

print("compute_event_mae() siap.")
````

### CELL 2.17 — Hitung 20 Gait Parameters

17 parameter semiogram + 3 tambahan (step count, uturn count, pola uturn).
Setiap parameter yang tidak bisa dihitung → `np.nan` (tidak error).


````python
# ════════════════════════════════════════════════════════════
#  SIGNAL ANALYSIS HELPERS
# ════════════════════════════════════════════════════════════

def autocorr_norm(x: np.ndarray, max_lag: int = None) -> np.ndarray:
    """Normalized autocorrelation (lag 0 = 1.0)."""
    x  = x - x.mean()
    n  = len(x)
    ac = np.correlate(x, x, mode='full')[n-1:]
    ac = ac / (ac[0] + 1e-12)
    return ac[:max_lag+1] if max_lag is not None else ac

def harmonic_ratio(x: np.ndarray, freq: float,
                   stride_time: float, n_harmonics: int = 10) -> float:
    """
    Harmonic ratio: rasio amplitudo harmonic genap / ganjil
    dari FFT sinyal akselerasi.
    """
    try:
        if stride_time <= 0 or len(x) < 10: return np.nan
        f0     = 1.0 / stride_time
        fft_m  = np.abs(np.fft.rfft(x))
        freqs  = np.fft.rfftfreq(len(x), d=1.0/freq)
        even, odd = 0.0, 0.0
        for k in range(1, n_harmonics + 1):
            idx = int(np.argmin(np.abs(freqs - k * f0)))
            if k % 2 == 0: even += fft_m[idx]
            else:           odd  += fft_m[idx]
        return float(even / (odd + 1e-12))
    except Exception:
        return np.nan

def ldlj_acc(acc_mag: np.ndarray, freq: float, duration: float) -> float:
    """
    Log Dimensionless Jerk dari magnitude akselerasi.
    Makin kecil (kurang negatif) = gerakan lebih smooth.
    """
    try:
        if duration <= 0 or len(acc_mag) < 3: return np.nan
        jerk     = np.diff(acc_mag) * freq
        jerk_int = np.trapz(jerk**2) / freq
        pk       = np.max(np.abs(acc_mag))
        if pk <= 0 or jerk_int <= 0: return np.nan
        return float(np.log((duration**3 * pk**2) / (2 * jerk_int + 1e-12)))
    except Exception:
        return np.nan

def sparc_rotation(gyr_mag: np.ndarray, freq: float,
                   fc: float = 10.0, padlevel: int = 4) -> float:
    """
    Spectral Arc Length dari magnitude gyroscope.
    Makin besar (kurang negatif) = putaran lebih smooth.
    """
    try:
        if len(gyr_mag) < 10: return np.nan
        N_pad = int(2 ** np.ceil(np.log2(len(gyr_mag) * 2**padlevel)))
        fm    = np.abs(np.fft.rfft(gyr_mag, n=N_pad))
        fm    = fm / (fm[0] + 1e-12)
        fr    = np.fft.rfftfreq(N_pad, d=1.0/freq)
        mask  = fr <= fc
        fm, fr = fm[mask], fr[mask]
        if len(fm) < 2: return np.nan
        df   = fr[1] - fr[0]
        deriv = np.diff(fm) / (df + 1e-12)
        return float(-np.trapz(np.sqrt(1 + (deriv/fc)**2), fr[:-1]))
    except Exception:
        return np.nan

# ════════════════════════════════════════════════════════════
#  TEMPORAL HELPERS
# ════════════════════════════════════════════════════════════

def stance_swing(hs: list, to: list, freq: float) -> tuple:
    """
    Stance time = HS → TO berikutnya (kaki yang sama).
    Swing  time = TO → HS berikutnya (kaki yang sama).
    """
    hs, to = sorted(hs), sorted(to)
    stances, swings = [], []
    for h in hs:
        nxt = [t for t in to if t > h]
        if nxt: stances.append((nxt[0] - h) / freq)
    for t in to:
        nxt = [h for h in hs if h > t]
        if nxt: swings.append((nxt[0] - t) / freq)
    return np.array(stances), np.array(swings)

def double_stance(lhs, lto, rhs, rto, freq):
    """Double stance: singkat setelah setiap HS sebelum TO berikutnya."""
    all_hs = sorted(lhs + rhs)
    all_to = sorted(lto + rto)
    dst = []
    for h in all_hs:
        nxt = [t for t in all_to if t > h]
        if nxt: dst.append((nxt[0] - h) / freq)
    arr = np.array(dst)
    return arr[arr > 0] if len(arr) else np.array([])

def classify_uturn(n_steps: int) -> str:
    """
    Klasifikasi pola U-turn berdasarkan jumlah langkah.
    Rule awal — dapat diganti sesuai definisi klinis:
      Pattern I   : ≤2 langkah
      Pattern II  : 3–4 langkah
      Pattern III : >4 langkah
    """
    if   n_steps <= 2: return 'Pattern_I'
    elif n_steps <= 4: return 'Pattern_II'
    else:              return 'Pattern_III'

# ════════════════════════════════════════════════════════════
#  MAIN: COMPUTE 20 GAIT PARAMETERS
# ════════════════════════════════════════════════════════════

def compute_gait_parameters(events: dict, meta: dict,
                             df: pd.DataFrame) -> dict:
    """
    Hitung 20 gait parameters dari events + meta + raw signal DataFrame.
    Return dict — isi np.nan jika tidak bisa dihitung.
    """
    freq    = float(meta.get('freq', 100.0))
    ubounds = meta.get('uturnBoundaries', None)
    protocol= meta.get('protocol', '')
    dist    = 20.0 if '20' in protocol else (10.0 if '10' in protocol else 20.0)

    LHS = events.get('LHS', [])
    LTO = events.get('LTO', [])
    RHS = events.get('RHS', [])
    RTO = events.get('RTO', [])
    all_hs = sorted(LHS + RHS)

    # ── Temporal ─────────────────────────────────────────────
    l_str = np.diff(sorted(LHS)) / freq if len(LHS) >= 2 else np.array([])
    r_str = np.diff(sorted(RHS)) / freq if len(RHS) >= 2 else np.array([])
    all_stride = np.concatenate([l_str, r_str])

    StrT   = float(np.mean(all_stride))   if len(all_stride) else np.nan
    CVStrT = float(np.std(all_stride) / (np.mean(all_stride)+1e-12) * 100)              if len(all_stride) >= 2 else np.nan

    l_stance, l_swing = stance_swing(LHS, LTO, freq)
    r_stance, r_swing = stance_swing(RHS, RTO, freq)
    all_stance = np.concatenate([l_stance, r_stance])
    all_swing  = np.concatenate([l_swing,  r_swing])

    dst_arr = double_stance(LHS, LTO, RHS, RTO, freq)
    dstT    = float(np.mean(dst_arr)) if len(dst_arr) else np.nan
    # dstT as % of StrT
    dstT_pct = (dstT / StrT * 100)                if (not np.isnan(dstT) and not np.isnan(StrT) and StrT > 0) else np.nan
    CVdstT  = float(np.std(dst_arr) / (np.mean(dst_arr)+1e-12) * 100)               if len(dst_arr) >= 2 else np.nan
    swTr    = float(np.mean(all_swing) / (np.mean(all_stance)+1e-12))               if (len(all_swing) and len(all_stance)) else np.nan

    # Average speed
    if len(all_hs) >= 2 and not np.isnan(StrT):
        t_total = (all_hs[-1] - all_hs[0]) / freq
        V = dist / t_total if t_total > 0 else np.nan
    else:
        V = np.nan

    # Stride length
    SteL = float(V * StrT) if (not np.isnan(V) and not np.isnan(StrT)) else np.nan

    # ── U-turn ───────────────────────────────────────────────
    if ubounds and len(ubounds) == 2:
        us, ue           = int(ubounds[0]), int(ubounds[1])
        UtrT             = (ue - us) / freq
        uturn_hs         = [h for h in all_hs if us <= h <= ue]
        straight_hs      = [h for h in all_hs if not (us <= h <= ue)]
        uturn_step_count   = len(uturn_hs)
        straight_step_count= len(straight_hs)
        uturn_pattern      = classify_uturn(uturn_step_count)
    else:
        UtrT = np.nan
        uturn_step_count    = np.nan
        straight_step_count = len(all_hs)
        uturn_pattern       = 'Unknown'

    # ── Signal-based parameters ──────────────────────────────
    lb_map = get_lb_axis_map(df)
    T_sig  = df.shape[0]
    duration = T_sig / freq

    def get_col(key):
        if key in lb_map and lb_map[key] in df.columns:
            return df[lb_map[key]].fillna(0).values.astype(np.float64)
        return None

    acc_cc = get_col('CC')
    acc_ml = get_col('ML')
    acc_ap = get_col('AP')
    gyr_x  = get_col('GyrX')
    gyr_y  = get_col('GyrY')
    gyr_z  = get_col('GyrZ')

    # Magnitude acceleration
    acc_avail = [a for a in [acc_cc, acc_ml, acc_ap] if a is not None]
    if len(acc_avail) >= 3:
        n = min(len(a) for a in acc_avail)
        acc_mag = np.sqrt(sum(a[:n]**2 for a in acc_avail))
    elif acc_avail:
        acc_mag = np.abs(acc_avail[0])
    else:
        acc_mag = None

    # LDLJa
    LDLJa = ldlj_acc(acc_mag, freq, duration) if acc_mag is not None else np.nan

    # SPARCrot
    if gyr_x is not None and gyr_y is not None and gyr_z is not None:
        n = min(len(gyr_x), len(gyr_y), len(gyr_z))
        gyr_mag = np.sqrt(gyr_x[:n]**2 + gyr_y[:n]**2 + gyr_z[:n]**2)
        SPARCrot = sparc_rotation(gyr_mag, freq)
    elif gyr_x is not None:
        SPARCrot = sparc_rotation(np.abs(gyr_x), freq)
    else:
        SPARCrot = np.nan

    # RMSaML
    RMSaML = float(np.sqrt(np.mean(acc_ml**2))) if acc_ml is not None else np.nan

    # Autocorrelation peaks from CC acceleration
    if acc_cc is not None and len(acc_cc) > 10:
        max_lag = min(int(freq * 3), len(acc_cc) - 1)
        ac      = autocorr_norm(acc_cc, max_lag=max_lag)
        peaks, _ = find_peaks(ac[1:], height=0)
        P1acc   = float(ac[peaks[0]+1]) if len(peaks) >= 1 else np.nan
        P2acc   = float(ac[peaks[1]+1]) if len(peaks) >= 2 else np.nan
        P1P2acc = float(P1acc/(P2acc+1e-12))                   if (not np.isnan(P1acc) and not np.isnan(P2acc)) else np.nan
    else:
        P1acc = P2acc = P1P2acc = np.nan

    # Harmonic ratios
    st = StrT if not np.isnan(StrT) else 1.0
    iHRaAP = harmonic_ratio(acc_ap, freq, st) if acc_ap is not None else np.nan
    iHRaCC = harmonic_ratio(acc_cc, freq, st) if acc_cc is not None else np.nan
    iHRaML = harmonic_ratio(acc_ml, freq, st) if acc_ml is not None else np.nan

    return {
        'V'                  : V,
        'StrT'               : StrT,
        'UtrT'               : UtrT,
        'LDLJa'              : LDLJa,
        'SPARCrot'           : SPARCrot,
        'CVStrT'             : CVStrT,
        'CVdstT'             : CVdstT,
        'P1acc'              : P1acc,
        'P2acc'              : P2acc,
        'SteL'               : SteL,
        'RMSaML'             : RMSaML,
        'iHRaAP'             : iHRaAP,
        'iHRaCC'             : iHRaCC,
        'iHRaML'             : iHRaML,
        'P1P2acc'            : P1P2acc,
        'swTr'               : swTr,
        'dstT'               : dstT_pct,
        'straight_step_count': straight_step_count,
        'uturn_step_count'   : uturn_step_count,
        'uturn_pattern'      : uturn_pattern,
    }

print("compute_gait_parameters() siap (20 parameter).")
````

### CELL 2.18 — Batch Inference Semua Trial

````python
def run_batch_inference(trial_list: list, model,
                         scaler, device=device,
                         save_pred_json=True) -> list:
    """
    Full pipeline inference untuk semua trials:
      load → preprocess → predict → extract events
      → compute gait params → collect MAE
    Jika satu trial gagal, log error dan lanjutkan.
    """
    results, errors = [], []

    for info in tqdm(trial_list, desc='Batch Inference'):
        tid = info['trial_id']
        try:
            # 1. Load & preprocess
            t    = load_trial(info)
            freq = float(t['meta'].get('freq', 100.0))
            X    = preprocess_signal(t['signal'], freq=freq)
            X    = apply_scaler(X, scaler)

            # 2. Full-trial prediction
            pred_labels, _ = predict_full_trial(model, X)

            # 3. Extract predicted events
            pred_ev = extract_events_from_sequence(pred_labels, freq=freq)

            # 4. Ground truth events & MAE
            gt_ev = parse_gt_events(t['meta'])
            mae   = compute_event_mae(pred_ev, gt_ev, freq=freq)

            # 5. Gait parameters
            gait_p = compute_gait_parameters(pred_ev, t['meta'], t['df'])

            # 6. Save per-trial prediction JSON
            if save_pred_json:
                out = {
                    'trial_id'       : tid,
                    'group'          : info['group'],
                    'freq'           : freq,
                    'pred_events_idx': {k:[int(x) for x in v]
                                        for k,v in pred_ev.items()},
                    'pred_events_sec': events_to_seconds(pred_ev, freq),
                    'gt_events_idx'  : {k:[int(x) for x in v]
                                        for k,v in gt_ev.items()},
                    'gt_events_sec'  : events_to_seconds(gt_ev, freq),
                    'mae_ms'         : {k:(float(v) if not np.isnan(v) else None)
                                        for k,v in mae.items()},
                }
                with open(os.path.join(PRED_DIR, f"{tid}_pred.json"), 'w') as f:
                    json.dump(out, f, indent=2)

            # 7. Build result row
            meta = t['meta']
            row  = {
                'subject'           : meta.get('subject', info['subject_id']),
                'group'             : info['group'],
                'pathologyKey'      : meta.get('pathologyKey', ''),
                'clinicalDeficitSide': meta.get('clinicalDeficitSide', ''),
                'session'           : meta.get('session', ''),
                'trial'             : meta.get('trial', ''),
                'freq'              : freq,
                'age'               : meta.get('age', np.nan),
                'gender'            : meta.get('gender', ''),
                'height'            : meta.get('height', np.nan),
                'weight'            : meta.get('weight', np.nan),
                'BMI'               : meta.get('BMI', np.nan),
                'FMA_LE'            : meta.get('evaluationScoreValue', np.nan),
                **gait_p,
                'n_LHS'             : len(pred_ev.get('LHS',[])),
                'n_LTO'             : len(pred_ev.get('LTO',[])),
                'n_RHS'             : len(pred_ev.get('RHS',[])),
                'n_RTO'             : len(pred_ev.get('RTO',[])),
                'MAE_LHS_ms'        : mae.get('LHS', np.nan),
                'MAE_LTO_ms'        : mae.get('LTO', np.nan),
                'MAE_RHS_ms'        : mae.get('RHS', np.nan),
                'MAE_RTO_ms'        : mae.get('RTO', np.nan),
                'MAE_overall_ms'    : mae.get('overall', np.nan),
            }
            results.append(row)

        except Exception as e:
            errors.append((tid, str(e)))

    if errors:
        print(f"\n[WARN] {len(errors)} trial gagal:")
        for tid, err in errors[:10]:
            print(f"  {tid}: {err}")

    return results

print("Menjalankan batch inference untuk semua 488 trials...")
ALL_RESULTS = run_batch_inference(ALL_TRIALS, model, GLOBAL_SCALER)
print(f"Selesai — {len(ALL_RESULTS)} trials diproses.")
````

1. **EVENT_TOLERANCE** (±5 sample): dikalibrasi per dataset — coba 3–7 sample
2. **U-turn pattern** (I/II/III): rule sederhana berbasis jumlah langkah, perlu validasi klinis
3. **Harmonic ratio** tidak stabil untuk trial sangat pendek atau event sangat sedikit
4. **SteL** diestimasi dari V×StrT (bukan dari motion capture)
5. **Class imbalance** event tinggi — pertimbangkan Focal Loss untuk iterasi berikutnya
